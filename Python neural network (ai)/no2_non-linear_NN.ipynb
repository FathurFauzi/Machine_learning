&emsp; That's why, to make the program can separate and process the right one, we need another neuron to make non-linear thing's possible.

&emsp; The difference between one linear neuron is we will use the Z output again to find another $y$ function, different weight and bias, then find the second sigmoid function:

$$ Z = \frac{1}{1+e^{-y}} $$

$$Y' = Z\cdot \omega_2 + b_2$$

$$ Z' = \frac{1}{1+e^{-y'}} $$

&emsp; And we will find the Forward Propagation's Value, but as in the 1 neural, one relearn isn't enough, so we need to do the Backpropagation to find the new **weight** and **bias** , to learn from its mistakes.

It's almost the same with the 1 neuron system, the difference is it gave another sigmoid derivations.

$$ d_2 = error_1 \cdot Z'\cdot (1-Z') $$

$Z\cdot (1-Z')$ as i mention at the previous file, it represent derivative of the sigmoid value that we search before.
Then find the first weight and bias's gradien as in the 1 neuron:

$$ \omega_2 = Z^T  \cdot d_2 $$ 
$$ b_1 = \Sigma d_2 $$

After that, we will calulate the $d_1$ using $d_2 \cdot \omega_2 $ as an error, and then calculate the sigmoid derivative of the previous layer (in this case is $Z$)

$$ d_1 = (d_2 \cdot \omega_2^T ) \cdot Z \cdot (1-Z) $$

And then find the first weight and bias:

$$ \omega_1(gradien) = X^T  \cdot d_1 $$ 
$$ b_1(gradien) = \Sigma d_1 $$

Then we update the weight and biases using our newfound gradient

$$ b_2 = b_2 - (lr * grad\_b_2) $$
$$ \omega_2 = \omega_2 - (lr * grad\_\omega_2) $$
$$ \omega_1 = \omega_1 - (lr * grad\_\omega_1) $$
$$ b_1 = b_1 - (lr * grad\_b_1) $$

**LR** is equals to **c** in our previous file, we customize it by our own. Then repeat the same process hundreds or thousands of time, and we will eventually find the best weight and bias to program it.

```python
import numpy as np

# Dataset XOR: tidak bisa dipisahkan satu garis lurus
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])
y = np.array([
    [0],
    [1],
    [1],
    [0]
])

np.random.seed(1)
lr = 0.5 # learning rate
epochs = 10000

# Inisialisasi bobot (kecil & acak) dan bias nol
W1 = np.random.randn(2, 2)   # input (2) -> hidden (2)
b1 = np.zeros((1, 2))
W2 = np.random.randn(2, 1)   # hidden (2) -> output (1)
b2 = np.zeros((1, 1))

def sigmoid(x):
    return 1.0 / (1.0 + np.exp(-x))

for epoch in range(1, epochs + 1):
    # ---- FORWARD ----
    H = sigmoid(X @ W1 + b1)          # 4x2
    Z = sigmoid(H @ W2 + b2)          # 4x1 ;   @ itu perkalian matrix

    # ---- BACKWARD ----
    d2 = (Z - y) * Z * (1 - Z)        # 4x1, error output
    grad_W2 = H.T @ d2                # 2x1
    grad_b2 = np.sum(d2, axis=0, keepdims=True)

    d1 = (d2 @ W2.T) * H * (1 - H)    # 4x2, error dikirim balik ke hidden
    grad_W1 = X.T @ d1                # 2x2
    grad_b1 = np.sum(d1, axis=0, keepdims=True)

    # ---- UPDATE ----
    W2 -= lr * grad_W2
    b2 -= lr * grad_b2
    W1 -= lr * grad_W1
    b1 -= lr * grad_b1

    if epoch in [1, 1000, 5000, 10000]:
        loss = np.mean((Z - y) ** 2)
        print(f"Epoch {epoch:<6} | loss = {loss:.6f}")

print("\nHasil akhir:")
for i in range(len(X)):
    print(f"Input {X[i]} -> prediksi {Z[i][0]:.4f} | target {y[i][0]}")
```

As you can see from the outputs, Now we can make the program show certain outputs, based of the input.

The concepts was still the same for more layer of This neural network, if we want to make more layer, just add another sigmoid equations that uses previous sigmoid equations (i mean to make another Z'', we can make Z' treated like X in first sigmoid equations). and then for the BackPropagations is the same as the second one, use d output before that multiplied by the current weight as an error.
