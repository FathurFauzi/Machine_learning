&emsp; According to **Sudy machine learning** There is mainly 3 function used in this neural network. One is **SIgmoid**, that we will learned, then there's **TanH (hyperbolic tangen)** and **ReLu (Rectified linear unit)**.

in a neuron, they have 2 essential equations, that is linear equations, and sigmoid equations. linear equations was like this:

$$ Y= X\cdot \omega + b $$
where:
- Y = the outputs
- X = inputs
- ω = weight (constant, gradient, show how important the input could change output)
- b = bias

and the other equations is **Sigmoid equations** . This equation makes a Y number that can be any number (from -∞ to ∞ ) into only a number from 0 to 1 (like a percentage in probability). The equation as follows: $$ Z = \frac{1}{1+e^{-y}} $$

when the **y** number is** negative**, it will go from  **0.5 aproach 0**, when the **y** value is **positive**, it will go from **0.5 aproach 1**.

This is only a half way of how computer thinks, now it only show a probability, now to make it can learn from it's mistakes, we need to know another equations:

To actually know how wrong we are, we need to compare our result with the actual result right? with same idea, we will make computer know how wrong they are by comparing this **Z** to the expected value, using simple differences equations:

$$ error = Z- y_{true}$$

Now we know the error, but to know excactly how far the predictions is wrong, we use MSE (Mean Squared Error). as it said, we will squared the mean (average) of the error, so the network calculates the MSE to see how far off its predictions are from the actual targets.

$$(\text{MSE}=\frac{1}{n}\sum _{i=1}^{n}(y_{i}-{\hat y}_{i})^{2})$$

where:

• n: Total number of data samples.

• $ y_{i} $: The actual ground-truth value.

• $\hat {y}_{i}$: The predicted value generated by the neural network.

this is only for us to see the error, but for the computer, it's not important yet.

Our next step is multiply the error with the derivative of our **Sigmoid value**. the derivative of **Sigmoid value** is as follows:

$$ Z' = Z ⋅ (1 - Z) $$

now multiply **z'** with error to find the derivative

$$ D_{value} = error  ⋅ Z'$$

with $ D_{value}$ , the computer can finally decide the correct $ω$ and  $b$ value by process it to some values:
- to calculate new $ω$ ,  first we need to multiply the **Transpos value** of the input $(X)$ with the $ D_{value}$ to find a **Gradient of ω**, and then we will subtract already the existing $ω$ with its gradient. we can multiply the gradient with some constants, so the correction can go smoother

$$ ω_{grad} = X^T⋅ D_{value} $$

$$ ω' = ω - (c \cdot ω_{grad})$$

In neural network, we call **c** as a learning rate, the bigger the value (aproaching 1), the bigger the change, but like when we riding a bike in an almost straight road, we didn't need to steer it to hard, if we did, we will zigzag and never settle on the straight line, same as this computer learning.

- now for the $b$ value, we just need to sum all of the already found $ D_{value}$  to found $b_{grad}$, and same as  $ω$, we can find $b'$ by subtracting the already exist $b$ with $(b_{grad}⋅c)$

$$ b_{grad} = ΣD_{value} $$

$$ b' = b- (c\cdot b_{grad}) $$

now we can found the new $ω$ and $b$, we just need to repeat the same process **thousands of time**, until the value is close with the expected one.

You can try by your own using this code below.

```python
import numpy as np

# 1. Dataset Gerbang Logika OR
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Target aktual:  NAND GATE
y_true = np.array([[1], [1], [1], [0]])

# 2. Inisialisasi Bobot & Bias
np.random.seed(42)
w = np.random.randn(2, 1)  # Bobot acak awal
b = 0.0                    # Bias awal
learning_rate = 0.5
training_count = 1500
def sigmoid(y):
    return 1.0 / (1.0 + np.exp(-y))

def sigmoid_derivative(z):
    return z * (1.0 - z)

print("=== Kondisi Awal Sebelum Belajar ===")
print(f"first weight w: [{w[0][0]:.3f}, {w[1][0]:.3f}], First Bias b: {b:.3f}")
pred_awal = sigmoid(np.dot(X, w) + b)
for i in range(len(X)):
    print(f"  Input: {X[i]} -> first prediction : {pred_awal[i][0]:.4f} (Target: {y_true[i][0]})")

print("\n=== Proses Belajar (Training Loop) ===")
# print monitoring table's header
print(f"{'Epoch':<8} | {'Total Loss (MSE)':<18} | {' w1':<10} | {' w2':<10} | {'Bias b':<10}")
print("-" * 65)

for epoch in range(1, training_count+1):
    # A. FORWARD PASS
    y_linier = np.dot(X, w) + b
    z_prediksi = sigmoid(y_linier)

    # B. calculate ERROR / LOSS (Mean Squared Error)
    error = z_prediksi - y_true
    loss = np.mean(error ** 2)

    # C. BACKPROPAGATION
    d_output = error * sigmoid_derivative(z_prediksi)
    grad_w = np.dot(X.T, d_output)
    grad_b = np.sum(d_output)

    # D. UPDATE weight (Gradient Descent)
    w -= learning_rate * grad_w
    b -= learning_rate * grad_b

    # show learning proccess in some intervals
    if epoch in [1, 10, 50,67, 100, 300, 500, 1000, 1200, 1500]:
        print(f"{epoch:<8} | {loss:<18.6f} | {w[0][0]:<10.4f} | {w[1][0]:<10.4f} | {b:<10.4f}")

print(f"\n=== after {training_count} Epoch ===")
pred_akhir = sigmoid(np.dot(X, w) + b)
for i in range(len(X)):
    val = pred_akhir[i][0]
    print(f"Input: {X[i]} -> Prediksi: {val:.4f} (Dibulatkan: {int(round(val))}) | Target: {y_true[i][0]}")
```

As you can se at the output, first output, the value is not close to what we expect, you can clearly see it when you round the prediction output. but when we put it in the trainging loop, we can see that the weight and bias is slowly change to certain value.

At this moment, i make it trained 1500 times. **Prediksi** is a sum of $w_1×X_1$, $w_2 × X_2$, and $b$. While it's not 100% correct, we can see that the total loss of the last epoch is much lower than the first one. And when we round the output value, it's actually correct.

But this program only works in a linear input. if we make the $y_{true}$ on a completely random number it wont show a correct prediction, even when you trained them milion times. for example, i used a xor ouput [0,1,1,0], to trained, in the program below:
